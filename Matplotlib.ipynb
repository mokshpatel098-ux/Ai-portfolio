```python
import matplotlib.pyplot as plt
```

```python
import pandas as pd
```

```python
students = pd.DataFrame({
    "Students":["John","Mary","Ali","Grace",
                "James","Faith","Brian","lucy"],
    "Age":[20,21,22,20,23,21,22,20],
    "Gender":["Male","Female","Male","Female",
              "Male","Female","Male","Female"],
    "Math":[78,85,90,66,74,88,82,91],
    "Science":[80,79,95,70,88,90,76,94],
    "English":[75,88,92,72,81,86,79,90]
})
```

```python
print(students)
```

```python
#A line chart shows trends over time or ordered data.
plt.plot(students["Students"],students["Math"])
plt.show()
```

```python
#Adding titles
plt.plot(students["Students"],students["Math"])
plt.title("Math Scores")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

```python
#Adding titles
plt.plot(students["Students"],students["Science"])
plt.title("Science Scores")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```
