Data frame is a two dimensional tables it contains rows and columns.Similar to an excel spreadsheet

```python
data = {
    "Name" : ["salma", "Babu", "Nand","irfan"],
    "Math" : [78, 34, 29 ,33],
    "English" : [58, 74, 69 ,93],
    "Science" : [68, 54, 59 ,91],
}
students = pd.DataFrame(data)
print(students)
```

```python
print(students.dtypes) #checking dta types
print(students.shape) #checking the shape of data frame
```
